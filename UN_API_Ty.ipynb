```python
endpoint = 'https://api.worldbank.org/v2/country'
```

```python
params = {
    'format': 'json',
    'per_page': 296
}
```

```python
response = requests.get(endpoint, params = params)
response
```

```python
res = response.json()
res
```

```python
country_df = pd.DataFrame(res[1])
country_df.head()
```

```python
country_df = pd.json_normalize(res[1])
country_df.head()
```

```python
country_df = country_df.drop(columns = ['id', 'name', 'region.id', 'region.iso2code', 'adminregion.id', 'adminregion.iso2code', 'incomeLevel.id', 'incomeLevel.iso2code', 'lendingType.id', 'lendingType.iso2code'])
country_df.head()
```

```python
country_df = country_df.rename(columns = {'iso2Code': 'country.id'})
country_df.head()
```

```python
full_df = merged_df.merge(country_df, on = ['country.id'], how = 'left')
```

```python
full_df.shape
```

```python
full_df['region.value'].value_counts(dropna = False)
```

```python
countries_df = full_df[full_df['region.value'] != 'Aggregates']
```

```python
countries_df.shape
```

```python
countries_df.head()
```

```python
countries_df['adminregion.value'].value_counts(dropna = False)
```
